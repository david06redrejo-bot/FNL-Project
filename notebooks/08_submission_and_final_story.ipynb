## Complete Project Progression

Our work followed the path we wanted the repository to tell:

```text
EDA -> preprocessing -> survey -> baselines -> RoBERTa CLS -> RoBERTa mean
-> improvements -> evaluation -> final submission
```

- **EDA:** we analyzed data files, schemas, labels, imbalance, duplicates, text patterns, and train/leaderboard length shift.
- **Preprocessing:** we chose light required cleaning because Spanish biomedical literals can lose meaning if we lowercase, strip accents, or remove punctuation too aggressively.
- **Survey grounding:** after reading ICD coding work such as Yan et al. (2022), we understood why ICD coding is not ordinary text classification: hierarchy, medical terminology, imbalance, and interpretability all matter.
- **Baselines:** majority, TF-IDF character n-grams, TF-IDF word n-grams, and retrieval gave us sanity checks before deep learning.
- **RoBERTa:** we tested CLS and mean pooling with a Spanish biomedical-clinical backbone.
- **Improvements:** class weighting, focal loss, hyperparameter tuning, safe data strategies, and ensembling were tested only after the basic models were understood.
- **Evaluation:** the final notebook compared all models, per-class metrics, confusion patterns, and confidence/error examples.
- **Submission:** `v09_ensemble` is kept as the best validation candidate, while `v10_vote_diverse_no_retrieval` is copied to `submissions/final_submission.csv` because it achieved the best verified Kaggle public score.

## Final Model Selection

The final decision has two useful signals. In the broader v10 search, `v10_vote_diverse_no_retrieval_val_weighted` reached the best validation accuracy, while `v10_vote_diverse_no_retrieval` achieved the best verified Kaggle public score.

Why this model?

- `v10_vote_diverse_no_retrieval_val_weighted` has the best validation accuracy from the completed v10 ensemble search: **0.583577**.
- `v10_vote_diverse_no_retrieval` has the best verified Kaggle public score: **0.587**.
- `v09_ensemble` remains important because it has the strongest macro F1 among the main completed candidates, which matters for imbalanced ICD categories.
- The final upload candidate is therefore `v10_vote_diverse_no_retrieval`, while `v09_ensemble` is kept as a macro-F1-oriented validation result and ensemble ablation.
- Error analysis still shows remaining limitations: short literals, ambiguity, abbreviations, rare categories, and broad ICD prefixes.

## Final Reflection

The most important thing we learned is that this task is small in input length but not simple. That is why the final reflection is not only about the leaderboard: it is about learning to connect data analysis, language, models, and responsibility. The literals are short, clinical, sometimes ambiguous, and imbalanced across categories. Strong performance came from respecting the data first, avoiding unsafe preprocessing, grounding decisions in the ICD coding survey, and only then combining models.

The final model is not perfect. It still makes errors on ambiguous short literals, abbreviations, rare categories, and broad ICD-prefix confusions. But it is traceable: every step from EDA to final submission is documented, reproducible, and connected to what we learned during the course.
