## Complete Project Progression

Our work followed the path we wanted the repository to tell:

```text
EDA -> preprocessing -> survey -> baselines -> RoBERTa CLS -> RoBERTa mean
-> improvements -> evaluation -> final submission
```

- **EDA:** we analyzed data files, schemas, labels, imbalance, duplicates, text patterns, and train/leaderboard length shift.
- **Preprocessing:** we chose light required cleaning because Spanish biomedical literals can lose meaning if we lowercase, strip accents, or remove punctuation too aggressively.
- **Survey grounding:** after reading ICD coding work such as Yan et al. (2022), we understood why ICD coding is not ordinary text classification: hierarchy, medical terminology, imbalance, and interpretability all matter.
- **Baselines:** majority, TF-IDF character n-grams, TF-IDF word n-grams, and retrieval gave us sanity checks before deep learning.
- **RoBERTa:** we tested CLS and mean pooling with a Spanish biomedical-clinical backbone.
- **Improvements:** class weighting, focal loss, hyperparameter tuning, safe data strategies, and ensembling were tested only after the basic models were understood.
- **Evaluation:** the final notebook compared all models, per-class metrics, confusion patterns, and confidence/error examples.
- **Submission:** `v09_ensemble` is declared as the final candidate and copied to `submissions/final_submission.csv`.

## Final Reflection

The most important thing we learned is that this task is small in input length but not simple. The literals are short, clinical, sometimes ambiguous, and imbalanced across categories. Strong performance came from respecting the data first, avoiding unsafe preprocessing, grounding decisions in the ICD coding survey, and only then combining models.

The final model is not perfect. It still makes errors on ambiguous short literals, abbreviations, rare categories, and broad ICD-prefix confusions. But it is traceable: every step from EDA to final submission is documented, reproducible, and connected to what we learned during the course.
